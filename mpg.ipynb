```python
import seaborn as sns 
mpg = sns.load_dataset("mpg")
```

```python
sns.relplot(data=mpg, x='horsepower', y='mpg')
```

```python
sns.catplot(data=mpg, x='origin', y='mpg', kind='box')
```

```python
sns.relplot(data=mpg, x='model_year', y='mpg', kind='line', errorbar=None)
```

```python
sns.displot(data=mpg, x='weight', y='mpg', cbar=True)
```

```python
sns.relplot(data=mpg, x='acceleration', y='mpg', hue='cylinders')
```

```python
# With the data set provided we can conclude a variety of different facts regarding MPG. 
# First we can see that with more horsepower the vehicle's MPG will be impacted and start to decrease. 
# Second we see that Japan manufactures the vehicles with the highest rated MPG followed by Europe in second and USA last.
# We then see how as model years became newer they also came with better MPG overall as time went on. 
# The weight of the vehicle is also a factor on the MPG performance with vehicles with greater than 3000lbs 
# impacted the most in loss of MPG.
# Lastly, we can conclude that acceleration alone doesn't direclty impact MPG but rather the number of cylinders in the vehicle. 
# 8 cylinders having the most power output needing the most gasoline places the lowest amongst its counterparts in terms of MPG. 
```
